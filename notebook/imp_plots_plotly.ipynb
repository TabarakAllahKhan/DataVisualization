```python
import plotly.express as px
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df=sns.load_dataset('iris')
df.head()
```

## 1. Scatter Plot

```python
fig1=px.scatter(df,x='sepal_length',y='sepal_width',color='species')
fig1.show()

```

## 2. Line Chart

```python
# sort the data

df_line = df.sort_values(by=['sepal_length'])
fig2=px.line(df_line,x='sepal_length',y='sepal_width',color='species')
fig2.show()
```

## 3. Bar Chart

```python
df_bar=df.groupby(['species']).mean().reset_index()
df
```

```python
fig3=px.bar(df_bar,x='sepal_length',y='sepal_width')
fig3.show()
```

## 4.Box Plot

```python
fig4=px.box(df,x='species',y='sepal_width',color='species')
fig4.show()
```

## 5. Violen Plot

```python
fig5=px.violin(df,x='species',y='sepal_width',color='species')
fig5.show()
```

```python
fig5=px.violin(df,x='species',y='sepal_width',color='species',box=True)
fig5.show()
```

## 6. Histogram

```python
fig6=px.histogram(df,x='sepal_width')
fig6.show()
```
